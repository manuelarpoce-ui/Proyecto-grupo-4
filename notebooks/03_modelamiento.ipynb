# Proyecto Grupo 4: Predicción de incumplimiento de pagos de tarjetas de crédito

## 1. Introducción

En esta etapa se desarrollarán y entrenarán diferentes modelos de aprendizaje automático con el propósito de predecir si un cliente incumplirá el pago de su tarjeta de crédito el próximo mes.

Se trabajará con tres modelos de clasificación: modelo base (Baseline), regresión logística y árbol de decisión.

Los modelos utilizarán las variables predictoras identificadas durante el análisis exploratorio y los datos preparados en la etapa de procesamiento.

Finalmente, se generarán predicciones y se comparará el desempeño inicial de los modelos mediante métricas de clasificación, con especial atención a la capacidad de identificar correctamente los clientes que incumplen sus pagos.

```python
# ============================================================
# 2. IMPORTAR LIBRERÍAS Y FUNCIONES
# ============================================================

import pandas as pd
import numpy as np

from grupo_4.limpieza import ejecutar_limpieza

from grupo_4.entrenamiento import (
    separar_variables,
    dividir_datos,
    entrenar_modelos,
    ejecutar_entrenamiento
)

print("Librerías y funciones importadas correctamente.")
```

```python
# ============================================================
# 3. CARGAR LOS DATOS
# ============================================================

# Ejecutar el proceso de limpieza
df = ejecutar_limpieza()

# Consultar las dimensiones de la base
print("Dimensiones de la base de datos:", df.shape)

# Mostrar las primeras filas
df.head()
```

```python

# ============================================================
# 4. SEPARAR VARIABLES PREDICTORAS Y VARIABLE OBJETIVO
# ============================================================

# Separar las variables predictoras (X) y la variable objetivo (y)
X, y = separar_variables(df)

# Consultar dimensiones
print("Dimensiones de las variables predictoras (X):", X.shape)
print("Dimensiones de la variable objetivo (y):", y.shape)

# Mostrar las primeras filas de las variables predictoras
X.head()
```

```python

# ============================================================
# 5. DIVIDIR LOS DATOS EN ENTRENAMIENTO Y PRUEBA
# ============================================================

# Dividir los datos
X_train, X_test, y_train, y_test = dividir_datos(X, y)

# Consultar dimensiones de los conjuntos
print("Dimensiones de entrenamiento:")
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("\nDimensiones de prueba:")
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)

# Consultar distribución de la variable objetivo
print("\nDistribución de incumplimiento en entrenamiento:")
print(y_train.value_counts(normalize=True) * 100)

print("\nDistribución de incumplimiento en prueba:")
print(y_test.value_counts(normalize=True) * 100)
```

```python

# ============================================================
# 6. ENTRENAR LOS MODELOS
# ============================================================

# Ejecutar el proceso de entrenamiento
resultado_entrenamiento = ejecutar_entrenamiento(df)

# Recuperar los conjuntos de entrenamiento y prueba
X_train = resultado_entrenamiento["X_train"]
X_test = resultado_entrenamiento["X_test"]
y_train = resultado_entrenamiento["y_train"]
y_test = resultado_entrenamiento["y_test"]

# Recuperar los modelos entrenados
modelos = resultado_entrenamiento["modelos"]

# Consultar los modelos entrenados
print("\nModelos entrenados correctamente:")

for nombre in modelos.keys():
    print("-", nombre)
```

```python
# ============================================================
# 7. GENERAR PREDICCIONES
# ============================================================

# Generar predicciones para cada modelo
predicciones = {}

for nombre, modelo in modelos.items():
    predicciones[nombre] = modelo.predict(X_test)

# Consultar cantidad de predicciones por modelo
print("Cantidad de predicciones generadas:")

for nombre, prediccion in predicciones.items():
    print(f"{nombre}: {len(prediccion)}")
```

```python
# ============================================================
# 8. CONSULTAR DISTRIBUCIÓN DE LAS PREDICCIONES
# ============================================================

# Crear tabla con la distribución de predicciones por modelo
tabla_predicciones = pd.DataFrame({
    nombre: pd.Series(prediccion).value_counts().reindex([0, 1], fill_value=0)
    for nombre, prediccion in predicciones.items()
}).T

# Renombrar las columnas
tabla_predicciones.columns = ["No incumple (0)", "Incumple (1)"]

# Agregar el total de predicciones
tabla_predicciones["Total"] = (
    tabla_predicciones["No incumple (0)"] +
    tabla_predicciones["Incumple (1)"]
)

# Mostrar tabla
tabla_predicciones.style.format("{:,.0f}").set_caption(
    "Distribución de predicciones por modelo"
).set_properties(**{
    "text-align": "center",
    "padding": "8px"
}).set_table_styles([
    {"selector": "th", "props": [
        ("text-align", "center"),
        ("padding", "8px")
    ]}
])
```

### Análisis de la distribución de las predicciones

Al comparar la distribución de las predicciones, se observa que el modelo Baseline clasificó los 6.000 clientes como personas que no incumplen sus pagos, debido a que esta es la clase mayoritaria del conjunto de datos.

Por su parte, la regresión logística identificó 704 clientes como posibles incumplidores, mientras que el árbol de decisión identificó 627.

Estos resultados muestran que los dos modelos predictivos reconocen clientes pertenecientes a ambas clases. Sin embargo, esta distribución no permite determinar por sí sola cuál modelo tiene mejor desempeño. Para ello, será necesario comparar sus predicciones con los resultados reales mediante métricas de clasificación.

```python
# ============================================================
# 9. COMPARAR MÉTRICAS DE CLASIFICACIÓN
# ============================================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

# Crear diccionario para almacenar las métricas
resultados_modelos = {}

# Calcular métricas para cada modelo
for nombre, prediccion in predicciones.items():

    resultados_modelos[nombre] = {
        "Accuracy": accuracy_score(y_test, prediccion),
        "Precision": precision_score(y_test, prediccion, zero_division=0),
        "Recall": recall_score(y_test, prediccion, zero_division=0),
        "F1-score": f1_score(y_test, prediccion, zero_division=0)
    }

# Crear tabla comparativa
tabla_metricas = pd.DataFrame(resultados_modelos).T

# Mostrar resultados
tabla_metricas.style.format("{:.4f}").set_caption(
    "Comparación de métricas de clasificación"
).set_properties(**{
    "text-align": "center",
    "padding": "8px"
}).set_table_styles([
    {"selector": "th", "props": [
        ("text-align", "center"),
        ("padding", "8px")
    ]}
])
```

### Justificación de las métricas de evaluación

Para comparar el desempeño de los modelos se seleccionaron las métricas Accuracy, Precision, Recall y F1-score. Esta elección responde a las características del conjunto de datos, en el que el 77,88 % de los clientes no incumplieron sus pagos, mientras que el 22,12 % sí lo hicieron.

Debido a este desbalance, no es suficiente evaluar los modelos únicamente mediante el Accuracy, ya que podrían obtener resultados aparentemente buenos al clasificar correctamente la mayoría de los clientes de la clase más frecuente, sin identificar adecuadamente a quienes incumplen.

Por esta razón, se incluyeron las siguientes métricas:

- **Accuracy:** permite conocer la proporción total de clientes clasificados correctamente y ofrece una visión general del desempeño.
- **Precision:** permite medir qué proporción de los clientes identificados como posibles incumplidores realmente incumplió. Es importante para evitar generar demasiadas alertas incorrectas.
- **Recall:** permite identificar qué proporción de los clientes que realmente incumplieron fue detectada por el modelo. Es especialmente relevante porque el proyecto busca reconocer oportunamente los posibles incumplimientos.
- **F1-score:** combina Precision y Recall en una sola medida, permitiendo comparar el equilibrio entre la detección de incumplimientos y la cantidad de falsas alarmas.

Se dará especial importancia al Recall y al F1-score, debido a que el objetivo del proyecto no es solamente clasificar correctamente a la mayoría de los clientes, sino también identificar la mayor cantidad posible de aquellos que podrían incumplir sus obligaciones de pago.

**¿Por qué no se utiliza el coeficiente de determinación R²?**

No se utiliza el coeficiente de determinación R² porque este indicador se emplea principalmente en problemas de regresión, en los que se busca predecir valores numéricos continuos. En nuestro proyecto, el objetivo es clasificar a los clientes en dos categorías: incumple (1) o no incumple (0). Por esta razón, las métricas de clasificación como Accuracy, Precision, Recall y F1-score son más apropiadas para evaluar el desempeño de los modelos.
