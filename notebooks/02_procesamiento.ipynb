# Proyecto Grupo 4: Predicción de incumplimiento de pagos de tarjetas de crédito

## Procesamiento de los datos

En esta etapa se realizará la preparación de los datos para utilizarlos en los modelos de aprendizaje automático.

Se partirá de la base de datos obtenida después del proceso de limpieza y se organizarán las variables predictoras y la variable objetivo, posteriormente, se definirán los conjuntos de entrenamiento y prueba, y se aplicarán las transformaciones necesarias para preparar la información.

El procesamiento contempla la imputación de posibles valores faltantes, la estandarización de las variables numéricas y la codificación de las variables categóricas, mediante un pipeline que permitirá mantener un proceso organizado y consistente.

```python
# ============================================================
# 1. IMPORTAR LIBRERÍAS Y FUNCIONES
# ============================================================

import pandas as pd
import numpy as np

from grupo_4.limpieza import ejecutar_limpieza

from grupo_4.preprocesamiento import (
    VARIABLES_NUMERICAS,
    VARIABLES_CATEGORICAS,
    obtener_preprocesador
)

print("Librerías y funciones importadas correctamente.")
```

```python
# ============================================================
# 2. CARGAR LOS DATOS
# ============================================================

# Ejecutar el proceso de limpieza
df = ejecutar_limpieza()

# Consultar las dimensiones de la base
print("Dimensiones de la base de datos:", df.shape)

# Mostrar las primeras filas
df.head()
```

```python
# ============================================================
# 3. SEPARAR VARIABLES PREDICTORAS Y VARIABLE OBJETIVO
# ============================================================

from grupo_4.entrenamiento import separar_variables

# Separar las variables predictoras (X) y la variable objetivo (y)
X, y = separar_variables(df)

# Consultar dimensiones
print("Dimensiones de las variables predictoras (X):", X.shape)
print("Dimensiones de la variable objetivo (y):", y.shape)

# Mostrar las primeras filas de las variables predictoras
X.head()
```

```python
# ============================================================
# 4. IDENTIFICAR VARIABLES NUMÉRICAS Y CATEGÓRICAS
# ============================================================

# Consultar las variables numéricas
print("Variables numéricas:")
print(VARIABLES_NUMERICAS)

# Consultar cantidad de variables numéricas
print("\nCantidad de variables numéricas:", len(VARIABLES_NUMERICAS))

# Consultar las variables categóricas
print("\nVariables categóricas:")
print(VARIABLES_CATEGORICAS)

# Consultar cantidad de variables categóricas
print("\nCantidad de variables categóricas:", len(VARIABLES_CATEGORICAS))

# Verificar el total de variables
print("\nTotal de variables:", 
      len(VARIABLES_NUMERICAS) + len(VARIABLES_CATEGORICAS))
```

```python
# ============================================================
# 5. DIVIDIR LOS DATOS EN ENTRENAMIENTO Y PRUEBA
# ============================================================

from grupo_4.entrenamiento import dividir_datos

# Dividir los datos
X_train, X_test, y_train, y_test = dividir_datos(X, y)

# Consultar dimensiones de los conjuntos
print("Dimensiones de entrenamiento:")
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("\nDimensiones de prueba:")
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)

# Consultar distribución de la variable objetivo
print("\nDistribución de incumplimiento en entrenamiento:")
print(y_train.value_counts(normalize=True) * 100)

print("\nDistribución de incumplimiento en prueba:")
print(y_test.value_counts(normalize=True) * 100)
```

```python
# ============================================================
# 6. CONSTRUIR EL PREPROCESADOR
# ============================================================

# Obtener el preprocesador definido en el módulo
preprocesador = obtener_preprocesador()

# Consultar la estructura del preprocesador
print("Estructura del preprocesador:")
print(preprocesador)
```

```python
# ============================================================
# 7. APLICAR LAS TRANSFORMACIONES
# ============================================================

# Ajustar el preprocesador y transformar los datos de entrenamiento
X_train_procesado = preprocesador.fit_transform(X_train)

# Transformar los datos de prueba con los parámetros aprendidos
X_test_procesado = preprocesador.transform(X_test)

# Consultar las dimensiones de los datos transformados
print("Dimensiones después del procesamiento:")
print("X_train_procesado:", X_train_procesado.shape)
print("X_test_procesado:", X_test_procesado.shape)
```

- Se conservaron los registros: no se perdieron clientes durante el procesamiento.
- Pasamos de 23 a 87 columnas: esto ocurre porque las 9 variables categóricas se transforman mediante OneHotEncoder, que crea columnas binarias para representar sus categorías.
- Las variables numéricas se escalaron: quedaron en una escala comparable.
- No hubo fuga de información: el preprocesador aprendió los parámetros con los datos de entrenamiento y después aplicó esas mismas transformaciones a los datos de prueba.

### 7. Resultados del procesamiento

El preprocesamiento permitió preparar las variables para su utilización en los modelos de aprendizaje automático. Las 14 variables numéricas fueron imputadas mediante la mediana y estandarizadas, mientras que las 9 variables categóricas fueron imputadas mediante la categoría más frecuente y transformadas utilizando One-Hot Encoding.

Como resultado, las variables predictoras pasaron de 23 a 87 columnas, debido a la expansión de las categorías en variables binarias. Se conservaron los 24.000 registros de entrenamiento y los 6.000 registros de prueba.

Es importante destacar que el preprocesador se ajustó únicamente con los datos de entrenamiento y posteriormente se aplicó a los datos de prueba, evitando así la fuga de información.
